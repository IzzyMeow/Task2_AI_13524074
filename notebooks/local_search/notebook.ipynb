```python
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time
import random

from src.environment import generate_initial_state, calculate_cost
from src.algorithms import run_hill_climbing

random.seed(42)

initial_state = generate_initial_state()
initial_cost = calculate_cost(initial_state)

print(f"Cost Awal (Konflik Hard Constraint): {initial_cost}")
```

```python
variants = ["Basic", "Sideways Move", "Stochastic", "Random Restart"]
results = {}

for variant in variants:
    final_state, final_cost, history = run_hill_climbing(initial_state, variant=variant)
    results[variant] = {
        'final_cost': final_cost,
        'history': history,
        'total_iterations': len(history) - 1
    }
    print(f"[{variant}] Selesai dalam {len(history)-1} iterasi. Cost Akhir: {final_cost}")
```

```python
plt.figure(figsize=(12, 6))

for variant, data in results.items():
    iterations = [step['iteration'] for step in data['history']]
    costs = [step['cost'] for step in data['history']]
    
    plt.plot(iterations, costs, marker='o', label=f"{variant} (Akhir: {data['final_cost']})")

plt.title('Perbandingan Penurunan Cost: Hill-Climbing Variants')
plt.xlabel('Iterasi')
plt.ylabel('Total Konflik (Cost)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
def animate_search(history, algo_name):
    for step in history:
        clear_output(wait=True)
        
        iteration = step['iteration']
        cost = step['cost']
        state = step['state']
        restarts = step.get('restarts', 0)
        
        print("=== ASTRO-SCHEDULER ANIMATION ===")
        print(f"Algoritma : {algo_name} HC")
        print(f"Iterasi   : {iteration}")
        print(f"Cost      : {cost}")
        if restarts > 0:
            print(f"Restarts  : {restarts}")
        print("---------------------------------")
        
        for idx, mission in enumerate(state):
            print(f"[Misi {idx}] {mission['fleet']} -> {mission['planet']} (Hari {mission['day']})")
            
        time.sleep(0.3) 
        
    print("\n>> PENCARIAN SELESAI <<")
    if cost == 0:
        print("Status: SUCCESS (Global Optimum Ditemukan)")
    else:
        print("Status: FAILED (Terjebak di Local Optimum)")

# Jalankan animasi untuk salah satu varian (ubah string di bawah untuk melihat varian lain)
varian_animasi = "Random Restart"
animate_search(results[varian_animasi]['history'], varian_animasi)
```
